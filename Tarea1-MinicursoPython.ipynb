El siguiente programa le solicita al usuario ingresar una calificación que sea mayor o igual a 0 y menor o igual a 10, y la clasifica según el rango donde se encuentre, el programa devuelve la nota correspondiente con letra a la calificación, esto es: NA, S, B, MB.

```python
calificacion = float(input("Ingresa una calificación entre 0 y 10: "))

if calificacion < 0 or calificacion > 10:
    print("Calificación no válida")
elif calificacion < 6:
    print("La calificación ingresada corresponde a: NA")
elif calificacion < 7.6:
    print("La calificación ingresada corresponde a: S")
elif calificacion < 8.6:
    print("La calificación ingresada corresponde a: B")
else:
    print("La calificación ingresada corresponde a: MB")
```

En el ejemplo anterior, se le ingresó la calificación 9.7 y la salida fué: MB.

También podemos modificar el código ingresando un while, para meter una calificación tras otra, sin necesidad de ejecutar varias veces la celda.

```python
while True:
    entrada = input("Ingresa una calificación entre 0 y 10 (o escribe salir): ")

    if entrada == "salir":
        break

    calificacion = float(entrada)

    if calificacion < 0 or calificacion > 10:
      print("Calificación no válida")
    elif calificacion < 6:
      print("La calificación ingresada corresponde a: NA")
    elif calificacion < 7.6:
      print("La calificación ingresada corresponde a: S")
    elif calificacion < 8.6:
      print("La calificación ingresada corresponde a: B")
    else:
      print("La calificación ingresada corresponde a: MB")
```
