Lo primero que haremos es definir la función determinante, que toma una matriz y devuelve un número real. Definimos después la función regla_de_cramer, que recibe por argumentos A= la matriz de coeficientes del sistema y b= el vector de términos independientes.

```python
def determinante_3x3(M):
    return (
        M[0][0]*(M[1][1]*M[2][2] - M[1][2]*M[2][1])
        - M[0][1]*(M[1][0]*M[2][2] - M[1][2]*M[2][0])
        + M[0][2]*(M[1][0]*M[2][1] - M[1][1]*M[2][0])
    )

def regla_de_cramer_3x3(A, b):
    # Determinante principal
    D = determinante_3x3(A)

    if D == 0:
        print("El sistema no tiene solución única.")
        return

    # Matriz Dx
    Dx = [
        [b[0], A[0][1], A[0][2]],
        [b[1], A[1][1], A[1][2]],
        [b[2], A[2][1], A[2][2]]
    ]

    # Matriz Dy
    Dy = [
        [A[0][0], b[0], A[0][2]],
        [A[1][0], b[1], A[1][2]],
        [A[2][0], b[2], A[2][2]]
    ]

    # Matriz Dz
    Dz = [
        [A[0][0], A[0][1], b[0]],
        [A[1][0], A[1][1], b[1]],
        [A[2][0], A[2][1], b[2]]
    ]

    # Determinantes
    det_Dx = determinante_3x3(Dx)
    det_Dy = determinante_3x3(Dy)
    det_Dz = determinante_3x3(Dz)

    # Soluciones
    x = det_Dx / D
    y = det_Dy / D
    z = det_Dz / D

    print("Determinante principal D =", D)
    print("Dx =", det_Dx)
    print("Dy =", det_Dy)
    print("Dz =", det_Dz)

    print("\nSolución del sistema:")
    print("x =", x)
    print("y =", y)
    print("z =", z)

# Solución al ejercicio de tarea.
# Matriz de coeficientes
A = [
    [2, 4, 6],
    [4, 5, 6],
    [3, 1, -2]
]

# Vector de términos independientes
b = [18, 24, 4]

regla_de_cramer_3x3(A, b)
```
