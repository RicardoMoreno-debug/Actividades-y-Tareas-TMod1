El siguiente programa recibe un arreglo de números y lo que hace es sumar los números pares, los impares y te dice cuales son los números mayores a 20.

```python
numeros = [4, 7, 12, 15, 18, 21, 24, 30, 33, 40]

suma = 0
pares = 0
impares = 0
mayores_20 = []

for numero in numeros:
    print(numero)
    suma += numero

    if numero % 2 == 0:
        pares += 1
    else:
        impares += 1

    if numero > 20:
        mayores_20.append(numero)

print("Suma:", suma)
print("Pares:", pares)
print("Impares:", impares)
print("Mayores que 20:", mayores_20)
```

También podemos modificar el código para que le pida al usuario que ingrese los valores que quiera y con ellos haga exactamente lo mismo, abajo podemos ver el código.

```python
cantidad = int(input("¿Cuántos números quieres ingresar? "))

numeros = []

for i in range(1, cantidad + 1):
    numero = int(input("Ingresa un número: "))
    numeros.append(numero)

suma = 0
pares = 0
impares = 0
mayores_20 = []

for numero in numeros:
    print(numero)
    suma += numero

    if numero % 2 == 0:
        pares += 1
    else:
        impares += 1

    if numero > 20:
        mayores_20.append(numero)

print("Suma:", suma)
print("Pares:", pares)
print("Impares:", impares)
print("Mayores que 20:", mayores_20)
```
